```python
# 3 number problmes solve
import pandas as pd

data = {
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150]
}

df = pd.DataFrame(data)

# Columns
print("Columns:", df.columns)

# Index
print("Index:", df.index)

# Single Column
print("\nPrice Column:")
print(df['Price'])

# Full Row
print("\nRow 1:")
print(df.loc[1])
```

```python
# 4 problem solved

import pandas as pd

data = {
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150]
}

df = pd.DataFrame(data)
# First Row
print("First Row (iloc):")
print(df.iloc[0])

print("\nFirst Row (loc):")
print(df.loc[0])

# Second Column
print("\nSecond Column (iloc):")
print(df.iloc[:, 1])

print("\nSecond Column (loc):")
print(df.loc[:, 'Price'])

# 2x2 Middle Block
print("\n2x2 Block (iloc):")
print(df.iloc[1:3, 1:3])

print("\n2x2 Block (loc):")
print(df.loc[1:2, ['Price', 'Quantity']])
```

```python
import pandas as pd

# Create DataFrame
data = {
    'Name': ['Alamin', 'Rahim', 'Karim', 'Sakib'],
    'Math': [85, 90, 78, 92],
    'Science': [88, 76, 80, 95]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Drop one column
df_column_drop = df.drop('Science', axis=1)

print("\nAfter Dropping Column:")
print(df_column_drop)

# Drop one row
df_row_drop = df.drop(2)

print("\nAfter Dropping Row:")
print(df_row_drop)
```

```python
import pandas as pd

# Create DataFrame
data = {
    'Name': ['Alamin', 'Rahim', 'Karim', 'Sakib'],
    'Department': ['IT', 'HR', 'Finance', 'Marketing'],
    'Salary': [50000, 45000, 60000, 55000]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Sort by Salary (Ascending)
salary_sort = df.sort_values(by='Salary')

print("\nSorted by Salary (Ascending):")
print(salary_sort)

# Sort by Name (Descending)
name_sort = df.sort_values(by='Name', ascending=False)

print("\nSorted by Name (Descending):")
print(name_sort)
```
