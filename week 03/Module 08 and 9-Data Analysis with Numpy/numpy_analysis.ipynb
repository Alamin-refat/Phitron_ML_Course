# Ndarray

```python
import numpy as np

# 1D array
arr1 = np.array([1,2,3,4,5])

print(type(arr1))
print(arr1.ndim)

# 2D array
arr2 = np.array([[1,2,3], # Row1
                [4,5,6]   # Row2
                ])
print(arr2.ndim)

# 3D array
arr3 = np.array([
        # 1st floor
        [[1,2,3], # Row1
        [4,5,6] ],   # Row2
        # 2nd floor 
        [[1,2,3], # Row1
        [4,5,6] ]   # Row2
    
])
print(arr3.ndim)
```

## Array attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
# data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
# size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

# nd array data type

```python
arr = np.array([1,2,3])

print(arr.dtype)

arr = np.array([1.5,2.4,3.3])

print(arr.dtype)

```

```python
arr = np.array([1,2,3.3])
# upcasting
print(arr.dtype)
# upcasting to string
arr = np.array([1,2,3.3,'3'])
print(arr.dtype)
```

## selecting a data type for an array

```python
arr = np.array([1,2,3])
print(arr.dtype)
arr = arr.astype(np.int32)
print(arr.dtype)
print(arr)
```

# nd array creating from existing data

```python
# List
lst = [10,20,30,40,40.5]
arr = np.array(lst,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)

matrix = [[1,2,3], # Row1
         [4,5,6]   # Row2
                ]
arr = np.array(matrix,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# tuple
tpl=(10,20,30)
arr = np.array(tpl,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# set

st = {1,2,3}
arr = np.array(list(st))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# dictonary
dc = {'a' : 10,'b' : 20, 'c':30}

keys = dc.keys()
values = dc.values()
items = dc.items()
print(items)

# keys to ndarray
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# values to ndarray
arr = np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# array creation of random value

```python
## np.random.rand(shape)

arr = np.random.rand(2,3) # 0 -> 1

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
## np.random.randint(start,end,shape)

arr = np.random.randint(1,10,(10,10)) 

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
## np.random.uniform(start,end,shape)

arr = np.random.uniform(50,100,(3,3,4)) 

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# array creation with range

```python
# np.arange(first,end,step) > similar to range function

arr = np.arange(1,10,1).reshape(3,3)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start,end,koita value)

arr = np.linspace(1,10,10)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start,end,koita value)

arr = np.logspace(0,4,6,base=2)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# indexing and slicing

```python
# indexing

print(arr1)
print(arr2)

# 1d array indexing
print(arr1[2])

# 2d array indexing
print(arr2[0][2])
```

```python
# slicing
# 1d = arr[start:end:step]

arr1_mod=arr1[1:4]
print(arr1_mod)
```

```python
# 2d array slicing
#arr[row_start:row_end:step, column_start:column_end:step]

print(arr2)

# getting a row
row_0=arr2[0:1,]
print(row_0)

# geting a column
col_0=arr2[::,0:1]
print(col_0)
```

# practice problem solve

```python
# 1. You have a list of temperatures recorded in Celsius. Your task is to:
#Create a NumPy array from the given list.
#Convert all temperatures from Celsius to Fahrenheit. The formula is F = C* 1.8 + 32.
#Print the new Fahrenheit array.

import numpy as np
celsius_list =[10,20,30,40,50,60]
celcius_array = np.array(celsius_list)
farenheit_array = celcius_array * 1.8 + 32
print(farenheit_array)
```

```python
# 2.  You have a Python list containing the lap times (in seconds) for an athlete.

#Task:
#Convert the list into a NumPy array.
#Print the array's shape, size, and data type.
#The athlete wants to see their times in minutes. Create a new array by dividing all the lap times by 60.
#Print the new array of lap times in minutes.

lap_times = [60, 62, 58, 65, 70]

lap_array = np.array(lap_times)

print("Shape:", lap_array.shape)
print("Size:", lap_array.size)
print("Data type:", lap_array.dtype)

lap_minit = lap_array / 60

print("Lap times in minutes:", lap_minit)
```
