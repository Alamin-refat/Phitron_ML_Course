# ndarray

```python
arr = [1,2,3]

print(arr*5)

arr = [x*5 for x in arr]

print(arr)
```

```python
import numpy as np
arr1 = np.array([1,2,3])
arr1 = arr1*5
print(arr1)
```

# Ndarray

```python
import numpy as np

# 1D array
arr1 = np.array([1,2,3,4,5])

print(type(arr1))
print(arr1.ndim)

# 2D array
arr2 = np.array([[1,2,3], # Row1
                [4,5,6]   # Row2
                ])
print(arr2.ndim)

# 3D array
arr3 = np.array([
        # 1st floor
        [[1,2,3], # Row1
        [4,5,6] ],   # Row2

    
        # 2nd floor 
        [[1,2,3], # Row1
        [4,5,6] ]   # Row2
    
])

print(arr3.ndim)
```

## Array attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

# data type

print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

# size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

# nd array data type

```python
arr = np.array([1,2,3])

print(arr.dtype)

arr = np.array([1.5,2.4,3.3])

print(arr.dtype)

```

```python
arr = np.array([1,2,3.3])
# upcasting
print(arr.dtype)
# upcasting to string
arr = np.array([1,2,3.3,'3'])
print(arr.dtype)
```

## selecting a data type for an array

```python
arr = np.array([1,2,3])
print(arr.dtype)
arr = arr.astype(np.int32)
print(arr.dtype)
print(arr)
```

# nd array creating from existing data

```python
# List
lst = [10,20,30,40,40.5]
arr = np.array(lst,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)

matrix = [[1,2,3], # Row1
         [4,5,6]   # Row2
                ]
arr = np.array(matrix,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# tuple
tpl=(10,20,30)
arr = np.array(tpl,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# set

st = {1,2,3}
arr = np.array(list(st))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```
