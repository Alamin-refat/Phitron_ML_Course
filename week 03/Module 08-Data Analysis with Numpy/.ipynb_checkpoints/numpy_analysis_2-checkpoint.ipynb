# numpy array manupulation

```python
import numpy as np

arr = np.random.randint(1,100, size =(10,5)) #sample -10 , feature - 5
arr.shape
```

```python
b = arr.reshape(5,10)

b.shape
b.ndim
b
```

```python
flatten = b.flatten()

print(flatten.ndim)
flatten

column_wise_flattening=np.ravel(b,order='F') # row wise er jonno order='C' use korte hbe
print(column_wise_flattening)
```

```python
a = np.random.randint(1,10, size =(2,3))
b = np.random.randint(20,30, size =(2,3))

print(a)
print(b)
```

## concatanate

```python
con_row = np.concatenate((a,b),axis=0) # row_wise = 0
print(con_row)
con_col = np.concatenate((a,b),axis=1) # column_wise = 1
print(con_col)
# row wise concatanate korar somoye column same thakte hbe
# col wise concatenate korar  somoye row same thakte hbe
```

## transpose

```python
mat = np.array([[10,20,30],
                [40,50,60]
                ])
transpose = mat.T
print(transpose)
```

```python
# array split
a = np.random.randint(1,10, size =(10,))
print(a)
splitted_array = np.array_split(a,3)
print(splitted_array)

# split -> equal division
```

# arithmatic  operator and mathmatical function

```python
# faster for vectorization

x = np.array([10,20,30,40])
y = np.array([2,3,4,5])

add=np.add(x,y)
print(add)

sub=x-y
print(sub)

mul=x*y
print(mul)

div=x/y
print(div)

reminder=x%y
print(reminder)
```

```python
# mathmatical function

# trigronometri funtion
# sin,cos tan takes value radian

sin_val=np.sin(x)
print(sin_val)

deg_con=np.rad2deg(x)
print(deg_con)
```

# logical function

```python
# comparison operator - .>, >, >=, =<, ==, !=

x = np.array([10,20,30,40])
y = np.array([2,3,30,5])

greater = x>y
print(greater)
equal = x==y
print(equal)

# all() -> shob true kina
# any() -> jekono ekta true kina

print(np.all(greater))
print(np.any(equal))
```
