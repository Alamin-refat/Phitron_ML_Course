# Class and object

```python
class Phone:
    category= "Electronics"

    ## constructor

    def __init__(self,model,battery,camera,battery_percentage=100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage

    ## methods

    def charge(self, hour):
        print(f"Charge completed by {hour}")


    def capture(self,photo):
        if(self.battery_percentage)<=0:
            print("No charge")

        else:
            self.battery_percentage-=photo
            print(f"Photo captured in {self.model}")
        

# creating object

apple = Phone("iphone-17",4500,45)

bluebery = Phone("b-17",3500,40)
samsang = Phone("s-25",5000,50)
```

# Inheritance

```python
class Smartphone(Phone):
    def __init__(self,model,battery,camera,processor):
        super().__init__(model,battery,camera)
        self.processor = processor


    def charge(self,hour):
        print("Fast charging in precess......")
        super().charge(hour)


        

pro = Smartphone("X",6000,60,"SD")

print(pro.battery)

pro.charge(1)
```

# polymorphism

```python
class Camera:
    def __init__(self, name):
        self.name = name

    # method
    def cap(self):
        print("a photo is capture")



class Smart(Camera):
    def __init__(self,name,resulation):
        super().__init__(name)
        self.resulation = resulation

    # method overriding
    def cap(self):
        print("photo is capture by camera")

class Dslr(Camera):
    def __init__(self,name,resulation):
        super().__init__(name)
        self.resulation = resulation


    # method overriding
    def cap(self):
        print("photo is capture by dslr")


class Drone(Camera):
    def __init__(self,name,resulation):
        super().__init__(name)
        self.resulation = resulation


    # method overriding
    def cap(self):
        print("photo is capture by drone")



phone = Smart("iphone", 30)
dslr = Dslr("nikkon", 100)
drone = Drone("dji", 50)


phone.cap()
dslr.cap()
drone.cap()
```

# Encapsulation

```python
class Mobile:
    def __init__(self,name,model,imei):
        self.name = name
        self.model = model
        self.__imei = imei  #private

    def charge(self):
        print("Phone is charging........")

    # this method show imei
    def imei_getter(self):
        return self.__imei
        
iphone = Mobile("phone","Iphone","1xkaf")

print(iphone.name)
print(iphone.model)
print(iphone.imei_getter())

iphone.imei="123"

print(iphone.imei_getter())
```
