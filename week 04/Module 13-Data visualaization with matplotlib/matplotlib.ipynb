```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
```

# 2D Line Plot

```python
hours = [2 , 3 , 4, 1 , 5 , 7 ,3] 
days = [1 ,2 , 3 ,4 ,5 ,6 ,7]

plt.plot(days,hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Trend of a student study time over a week')

plt.plot(days,hours)
```

```python
hours_1 = [2 , 3 , 4, 1 , 5 , 7 , 3] 
hours_2 = [ 1 , 4 , 3 , 5 ,7 ,2 ,1]
days = [1 ,2 , 3 ,4 ,5 ,6 ,7] 

plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Trend of a student study time over a week')

plt.plot(days,hours_1)
plt.plot(days,hours_2)
```

# 2D line from pandas

```python
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'], df['Programming'])
```

```python
plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'], df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
# adding labels using plt.legend() 
plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming')
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing')

plt.legend(loc='best') 
```

```python
# adding some visualaization

plt.xlabel('Years') 
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming', color='red', linewidth=2)
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing', color='black', linewidth=2)

plt.legend() 
```
