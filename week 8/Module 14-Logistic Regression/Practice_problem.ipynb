## Problem 1
A logistic regression model predicts the probability of a student passing an exam:
Here x is hours studied.
a) What is the probability of passing if a student studies 2 hours?
b) What is the decision boundary value of x (where probability = 0.5)?

### solution

a) Calculation:$z = 0.5(2) + 0.2 = 1.0 + 0.2 = 1.2$$P(y=1|x=2) = \frac{1}{1 + e^{-1.2}}$Amra jani $e^{-1.2} \approx 0.3012$$P = \frac{1}{1 + 0.3012} = \frac{1}{1.3012} \approx 0.7686$ (ba $76.86\%$)Answer: The probability of passing is 0.7686 or 76.86%.

b) Calculation:$0.5x + 0.2 = 0$$0.5x = -0.2$$x = \frac{-0.2}{0.5} = -0.4$Answer: The decision boundary value of $x$ is $-0.4$. (Er mane jodi keu $-0.4$ ghonta pore, probality $0.5$ hobe. Bastobe shomoy negative hoy na, kintu mathematically eta-i boundary).

## Problem 2
Given a single data point: actual y = 1, predicted probability ŷ = 0.3.  
Calculate the Log Loss for this single example.

### solution
Calculation:Jehetu $y=1$, formula er ditiyo ongsho $(1-1)\log(1-\hat{y}) = 0$ hoye jabe.$\text{Loss} = -1 \cdot \ln(0.3)$  (Amra sadharonoto natural log $\ln$ use kori)$\ln(0.3) \approx -1.2039$$\text{Loss} = -(-1.2039) = 1.2039$Answer: The Log Loss for this single example is 1.2039.
