## Problem 1
A logistic regression model predicts the probability of a student passing an exam:
Here x is hours studied.
a) What is the probability of passing if a student studies 2 hours?
b) What is the decision boundary value of x (where probability = 0.5)?

### solution

a) Calculation:$z = 0.5(2) + 0.2 = 1.0 + 0.2 = 1.2$$P(y=1|x=2) = \frac{1}{1 + e^{-1.2}}$Amra jani $e^{-1.2} \approx 0.3012$$P = \frac{1}{1 + 0.3012} = \frac{1}{1.3012} \approx 0.7686$ (ba $76.86\%$)Answer: The probability of passing is 0.7686 or 76.86%.

b) Calculation:$0.5x + 0.2 = 0$$0.5x = -0.2$$x = \frac{-0.2}{0.5} = -0.4$Answer: The decision boundary value of $x$ is $-0.4$. (Er mane jodi keu $-0.4$ ghonta pore, probality $0.5$ hobe. Bastobe shomoy negative hoy na, kintu mathematically eta-i boundary).

## Problem 2
Given a single data point: actual y = 1, predicted probability ŷ = 0.3.  
Calculate the Log Loss for this single example.

### solution
Calculation:Jehetu $y=1$, formula er ditiyo ongsho $(1-1)\log(1-\hat{y}) = 0$ hoye jabe.$\text{Loss} = -1 \cdot \ln(0.3)$  (Amra sadharonoto natural log $\ln$ use kori)$\ln(0.3) \approx -1.2039$$\text{Loss} = -(-1.2039) = 1.2039$Answer: The Log Loss for this single example is 1.2039.

## Problem 3
You train a logistic regression model with weight w = 2.0  and bias b = -1.0.  
For a feature vector x = [1.5, 0.5],  
a) Compute z = w·x + b  
b) Compute the predicted probability ŷ = σ(z).  
c) If actual y = 1, should the model update w to increase or decrease? Brief some reasoning

### solution
) Compute $z=w\cdot x+b$Calculation:$z = (2.0 \times 1.5) + (2.0 \times 0.5) + (-1.0)$$z = 3.0 + 1.0 - 1.0 = 3.0$Answer: $z = 3.0
$b) Compute the predicted probability $\hat{y}=\sigma(z)$Calculation:$\hat{y} = \frac{1}{1 + e^{-3}} = \frac{1}{1 + 0.0498} = \frac{1}{1.0498} \approx 0.9526$Answer: $\hat{y} \approx 0.9526$
c)Answer: The model should increase $w$. Reasoning: Since the actual $y=1$ is greater than the predicted $\hat{y}=0.9526$, the model needs to output a higher probability. To increase the output of the sigmoid function, $z$ must increase. Since the input features $x$ are positive, increasing $w$ will increase $z$, pushing $\hat{y}$ closer to 1.

## Problem 5 
A logistic regression model is trained on a dataset with 2 features.  
The learned parameters are: w = [1.2, -0.8], b = 0.5.  
a) Write the equation of the decision boundary.  
b) A new point has x1 = 2.0, x2 = 1.5. Which class is predicted?  
c) How far (in terms of z value) is this point from the decision boundary?

### solution
a) The equation is 1.2x_1 - 0.8x_2 + 0.5 = 0 (Ba eta ke sajay likhle: 0.8x_2 = 1.2x_1 + 0.5  x_2 = 1.5x_1 + 0.625).
b) Calculation: Prothome $z$ ber kori:$z = 1.2(2.0) - 0.8(1.5) + 0.5 = 2.4 - 1.2 + 0.5 = 1.7$Jehetu $z = 1.7$ (jeta $0$ er cheye boro), tar mane probability $0.5$ er beshi hobe ($\sigma(1.7) > 0.5$).Answer: Class 1 is predicted.
c) The point is 1.7 units away from the decision boundary in terms of $z$ value.

## Problem 6
You are given a highly imbalanced dataset: 95% negative class (0) and 5% positive class (1).  
You train a logistic regression model and get 95% accuracy, but it predicts all samples as class 0.  
a) Why is accuracy misleading here?  
b) Which evaluation metric(s) would you use instead? 
c) How can you adjust the model to detect more positive samples?

### solution
a) Accuracy is misleading because the dataset is highly imbalanced. A dumb model that blindly predicts the majority class (0) for every sample will still achieve 95% accuracy, while completely failing to detect any sample from the minority class (1).
b) Precision: Model jader positive bolse, tader moddhe aslole koyjon positive.
Recall (Sensitivity): Total jotojon positive chilo, tader moddhe model koyjonke khuje ber korte parlo.
c) Class Weights Adjust kora: Model training er shomoy Class 1 er jonno Loss/Penalty bariye dewa (e.g., class_weight='balanced').
Threshold Shifting: Prediction classification threshold ta $0.5$ theke komiye $0.2$ ba $0.3$ e niye asha, jate model sahosh kore aro beshi positive predict kore.

## problem 7
Problem 7
Write pseudocode (or Python-like logic) for the following functions in logistic regression from scratch (no libraries except numpy for basic ops):  
a) sigmoid(z)
b) predict_probability(X, w, b)
c) compute_cost(X, y, w, b)
d) compute_gradients(X, y, w, b)
e) gradient_descent(X, y, w_init, b_init, alpha, iterations)

Focus on the loops and formulas, not exact Python syntax.

### Solutions

```python
## (a) sigmoid(z)
def sigmoid(z):
    return 1/(1+exp(-z))
```

```python
## b) predict_probability(X, w, b)
def predict_probability(X, w, b):
    # X: Matrix of features (shape: samples x features)
    # w: weights vector, b: bias scalar
    z = dot_product(X, w) + b
    return sigmoid(z)
```

```python
## c) compute_cost(X, y, w, b)
def compute_cost(X, y, w, b):
    m = length(y) # Total number of samples
    y_hat = predict_probability(X, w, b)
    
    # Log loss formula protita sample er jonno log loss er average
    total_loss = 0
    for i in range(m):
        loss_i = - (y[i] * log(y_hat[i]) + (1 - y[i]) * log(1 - y_hat[i]))
        total_loss += loss_i
        
    return total_loss / m
```

```python
## d) compute_gradients(X, y, w, b)
def compute_gradients(X, y, w, b):
    m = length(y)
    num_features = length(w)
    y_hat = predict_probability(X, w, b)
    
    # Error vector calculate kori
    error = y_hat - y
    
    # Gradients initialisation
    dw = array_of_zeros(num_features)
    db = 0
    
    # Loop protita feature er gradient calculation er jonno
    for j in range(num_features):
        sum_dw = 0
        for i in range(m):
            sum_dw += error[i] * X[i][j]
        dw[j] = sum_dw / m
        
    # Bias er gradient
    sum_db = 0
    for i in range(m):
        sum_db += error[i]
    db = sum_db / m
    
    return dw, db
```

```python
## e) gradient_descent(X, y, w_init, b_init, alpha, iterations)
def gradient_descent(X, y, w_init, b_init, alpha, iterations):
    w = w_init
    b = b_init
    
    for k in range(iterations):
        # 1. Gradient ber kora
        dw, db = compute_gradients(X, y, w, b)
        
        # 2. Parameters update kora (W = W - alpha * dW)
        w = w - alpha * dw
        b = b - alpha * db
        
        # Optional: majhe majhe cost print kore dekha loss komche kina
        if k % 100 == 0:
            current_cost = compute_cost(X, y, w, b)
            print("Iteration", k, "Cost:", current_cost)
            
    return w, b
```
