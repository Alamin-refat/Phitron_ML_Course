# Class and object

```python
class Phone:
    category= "Electronics"

    ##constructor

    def __init__(self,model,battery,camera):
        self.model = model
        self.battery = battery
        self.camera = camera

# creating object

apple = Phone("iphone-17",4500,45)

bluebery = Phone("b-17",3500,40)
samsang = Phone("s-25",5000,50)

print(apple.model)
print(bluebery.camera)
print(samsang.battery)
```
